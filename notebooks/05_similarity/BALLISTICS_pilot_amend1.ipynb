# Ballistics pilot, amendment 1: retrain failed models, re-gate (hard stop), strength-aware analysis

Follows `04_results/ballistics_pilot_plan_amendment1.json` (written before any retraining).

1. Any zoo model with validation AUC below 0.80 (expected: DINOv2 and CLIP ViT-B) is retrained with the full stabilized ViT recipe (layer-wise LR decay 0.65, 5-epoch warm-up, 15 epochs). Old short-recipe files are kept as `_shortrecipe`.
2. The calibration gate is recomputed on all 8 anchors, no exclusions. **If it fails, the notebook stops.**
3. Fingerprints for retrained models; analysis with strength handled explicitly (partial correlations, within-strength tertiles), leave-one-family-out, and the clinical test with 10% (primary) and 25% (secondary) tiers.

A100, High-RAM, `HF_TOKEN`. About 45 to 60 minutes.

```python
# Cell 4: training and prediction helpers
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import roc_auc_score
MEAN = torch.tensor([0.485, 0.456, 0.406]).view(1, 3, 1, 1); STD = torch.tensor([0.229, 0.224, 0.225]).view(1, 3, 1, 1)
TF_TRAIN = transforms.Compose([transforms.RandomResizedCrop(224, scale=(0.8, 1.0)), transforms.RandomHorizontalFlip(), transforms.RandomVerticalFlip(),
                               transforms.ColorJitter(0.2, 0.2, 0.1), transforms.ToTensor(), transforms.Normalize(MEAN.flatten().tolist(), STD.flatten().tolist())])
TF_EVAL = transforms.Compose([transforms.Resize(256), transforms.CenterCrop(224), transforms.ToTensor(), transforms.Normalize(MEAN.flatten().tolist(), STD.flatten().tolist())])
class DS(Dataset):
    def __init__(self, ids, labels, tf): self.ids, self.labels, self.tf = list(ids), list(labels), tf
    def __len__(self): return len(self.ids)
    def __getitem__(self, i): return self.tf(Image.open(INDEX[str(self.ids[i])]).convert('RGB')), int(self.labels[i]), i
def score(out): return torch.softmax(out.float(), 1)[:, 2:].sum(1)
@torch.no_grad()
def predict_ids(m, ids, bs=128):
    m.eval(); p = np.zeros(len(ids))
    for x, _, idx in DataLoader(DS(ids, [0] * len(ids), TF_EVAL), batch_size=bs, num_workers=NW, pin_memory=True):
        with torch.amp.autocast('cuda'): p[idx.numpy()] = score(m(x.to(dev))).double().cpu().numpy()
    return p
def train_zoo_model(name):
    r = RECIPE[Z.loc[name, 'recipe']]; random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED); torch.cuda.manual_seed_all(SEED)
    m = build(name).to(dev); tr, va = splits['train'], splits['val']
    g = torch.Generator(); g.manual_seed(SEED)
    dl_ = DataLoader(DS(tr.image_id.astype(str), tr.level, TF_TRAIN), batch_size=r['bs'], shuffle=True, drop_last=True, num_workers=NW, generator=g,
                     pin_memory=True, persistent_workers=True, prefetch_factor=4)
    opt = torch.optim.AdamW(m.parameters(), lr=r['lr'], weight_decay=r['wd']); E, W = r['epochs'], r['warmup']
    sch = torch.optim.lr_scheduler.LambdaLR(opt, lambda ep: (ep + 1) / W if ep < W else 0.5 * (1 + math.cos(math.pi * (ep - W) / max(1, E - W))))
    scaler = torch.amp.GradScaler('cuda'); lossf = nn.CrossEntropyLoss(); best = (-1, None, 0)
    for ep in range(E):
        m.train(); t0 = time.time()
        for x, y, _ in dl_:
            x, y = x.to(dev, non_blocking=True), y.to(dev, non_blocking=True); opt.zero_grad(set_to_none=True)
            with torch.amp.autocast('cuda'): loss = lossf(m(x), y)
            scaler.scale(loss).backward()
            if r['clip']: scaler.unscale_(opt); nn.utils.clip_grad_norm_(m.parameters(), r['clip'])
            scaler.step(opt); scaler.update()
        sch.step(); vauc = roc_auc_score((va.level >= 2).astype(int), predict_ids(m, va.image_id.astype(str)))
        if vauc > best[0]: best = (vauc, {k: v.detach().cpu().clone() for k, v in m.state_dict().items()}, ep + 1)
        print(f'   epoch {ep+1}/{E} val_auc {vauc:.4f} ({time.time()-t0:.0f}s)')
    m.load_state_dict(best[1]); return m, best[0], best[2]
print('ready')
```

```python
# Cell 5: amendment rule. Identify short-recipe failures and retrain them with the full stabilized ViT recipe
REC = f'{ZOO_OUT}/zoo_training_record.json'; rec = json.load(open(REC))
AMEND = f'{ZOO_OUT}/amendment1_record.json'; amend = json.load(open(AMEND)) if os.path.exists(AMEND) else {'retrained': {}}
FAILED = sorted(set([k for k, v in rec.items() if v['val_auc'] < 0.80 and not v.get('recipe') == 'full']) | set(amend['retrained']))
print('models below the 0.80 validation-AUC rule:', FAILED)
FULL_VIT = {'epochs': 15, 'lr': 3e-5, 'wd': 0.05, 'warmup': 5, 'clip': 1.0, 'bs': 32, 'lldecay': 0.65}
def find_blocks(m):
    best = None
    for _, mod in m.named_modules():
        if isinstance(mod, (nn.ModuleList, nn.Sequential)) and len(mod) >= 12 and (best is None or len(mod) > len(best)): best = mod
    return best
def llrd_optimizer(m, r):
    blocks = find_blocks(m); assert blocks is not None, 'no transformer blocks found'
    n = len(blocks); lr, d = r['lr'], r['lldecay']; named = list(m.named_parameters()); bid = {id(p) for b in blocks for p in b.parameters()}
    embed = [p for nm, p in named if id(p) not in bid and nm.startswith(('cls_token', 'pos_embed', 'patch_embed', 'reg_token', 'mask_token'))]
    head = [p for nm, p in named if id(p) not in bid and nm.startswith('head')]
    used = bid | {id(p) for p in embed} | {id(p) for p in head}; rest = [p for nm, p in named if id(p) not in used]
    groups = [{'params': embed, 'lr': lr * d ** n}] + [{'params': list(b.parameters()), 'lr': lr * d ** (n - 1 - i)} for i, b in enumerate(blocks)]
    groups += [{'params': rest, 'lr': lr}, {'params': head, 'lr': lr}]; groups = [g_ for g_ in groups if g_['params']]
    assert sum(p.numel() for g_ in groups for p in g_['params']) == sum(p.numel() for p in m.parameters())
    return torch.optim.AdamW(groups, lr=lr, weight_decay=r['wd'])
def train_full_vit(name):
    r = FULL_VIT; random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED); torch.cuda.manual_seed_all(SEED)
    m = build(name).to(dev); tr, va = splits['train'], splits['val']; g = torch.Generator(); g.manual_seed(SEED)
    dl_ = DataLoader(DS(tr.image_id.astype(str), tr.level, TF_TRAIN), batch_size=r['bs'], shuffle=True, drop_last=True, num_workers=NW, generator=g,
                     pin_memory=True, persistent_workers=True, prefetch_factor=4)
    opt = llrd_optimizer(m, r); E, W = r['epochs'], r['warmup']
    sch = torch.optim.lr_scheduler.LambdaLR(opt, lambda ep: (ep + 1) / W if ep < W else 0.5 * (1 + math.cos(math.pi * (ep - W) / max(1, E - W))))
    scaler = torch.amp.GradScaler('cuda'); lossf = nn.CrossEntropyLoss(); best = (-1, None, 0)
    for ep in range(E):
        m.train(); t0 = time.time()
        for x, y, _ in dl_:
            x, y = x.to(dev, non_blocking=True), y.to(dev, non_blocking=True); opt.zero_grad(set_to_none=True)
            with torch.amp.autocast('cuda'): loss = lossf(m(x), y)
            scaler.scale(loss).backward(); scaler.unscale_(opt); nn.utils.clip_grad_norm_(m.parameters(), r['clip']); scaler.step(opt); scaler.update()
        sch.step(); vauc = roc_auc_score((va.level >= 2).astype(int), predict_ids(m, va.image_id.astype(str)))
        if vauc > best[0]: best = (vauc, {k: v.detach().cpu().clone() for k, v in m.state_dict().items()}, ep + 1)
        print(f'   epoch {ep+1}/{E} val_auc {vauc:.4f} ({time.time()-t0:.0f}s)')
    m.load_state_dict(best[1]); return m, best[0], best[2]
for name in FAILED:
    if name in amend['retrained']: print('already retrained:', name); continue
    for old, new in ((f'{ZOO_CK}/{name}.pt', f'{ZOO_CK}/{name}_shortrecipe.pt'), (f'{ZOO_OUT}/preds/{name}_fresh.csv', f'{ZOO_OUT}/preds/{name}_shortrecipe_fresh.csv'),
                     (f'{ZOO_OUT}/fingerprints/{name}.npz', f'{ZOO_OUT}/fingerprints/{name}_shortrecipe.npz')):
        if os.path.exists(old) and not os.path.exists(new): os.rename(old, new)
    print(f'=== retraining {name} with the full stabilized ViT recipe ==='); t0 = time.time()
    m, vauc, ep = train_full_vit(name)
    torch.save({'state_dict': m.state_dict(), 'name': name, 'val_auc': vauc, 'best_epoch': ep, 'recipe': 'full_stabilized_vit'}, f'{ZOO_CK}/{name}.pt')
    p = predict_ids(m, FR.image_id); pd.DataFrame({'image_id': FR.image_id, 'predicted_probability': p}).to_csv(f'{ZOO_OUT}/preds/{name}_fresh.csv', index=False, float_format='%.17g')
    amend['retrained'][name] = {'short_recipe': rec[name], 'val_auc': round(float(vauc), 4), 'best_epoch': ep,
                                'fresh_auc': round(float(roc_auc_score(FR.binary_label, p)), 4), 'minutes': round((time.time() - t0) / 60, 1), 'recipe': 'full'}
    rec[name] = {**amend['retrained'][name], 'recipe': 'full'}; rec[name].pop('short_recipe')
    json.dump(amend, open(AMEND, 'w'), indent=2); json.dump(rec, open(REC, 'w'), indent=2); print('   done', amend['retrained'][name]); del m; torch.cuda.empty_cache()
still_bad = [k for k, v in rec.items() if v['val_auc'] < 0.80]
print('models still below 0.80 after the rule:', still_bad or 'none')
```

```python
# Cell 6: outcomes and the re-computed calibration gate (all 8 anchors, no exclusions; HARD STOP on failure)
from scipy.stats import spearmanr, rankdata
G = FR.level.to_numpy(); n = len(G); referable, sev, pdr = G >= 2, G >= 3, G == 4
def refer(s, f=0.25):
    k = int(round(f * len(s))); r = np.zeros(len(s), bool); r[np.argsort(-s, kind='stable')[:k]] = True; return r
S = {nm: pd.read_csv(f'{ZOO_OUT}/preds/{nm}_fresh.csv', dtype={'image_id': str}).set_index('image_id').predicted_probability.reindex(FR.image_id).to_numpy() for nm in Z.index}
REF = {nm: refer(s) for nm, s in S.items()}; RK = {nm: rankdata(s) / n for nm, s in S.items()}
def comiss(ra, rb, tgt):
    a, b = tgt & ~ra, tgt & ~rb; T = tgt.sum()
    return ((a & b).sum() / T) / ((a.sum() / T) * (b.sum() / T)) if a.sum() and b.sum() else np.nan
NAMES = list(Z.index); PAIRS = list(itertools.combinations(NAMES, 2))
OUT = pd.DataFrame([{'a': a, 'b': b, 'comiss_sev': comiss(REF[a], REF[b], sev), 'comiss_pdr': comiss(REF[a], REF[b], pdr),
                     'comiss_ref': comiss(REF[a], REF[b], referable)} for a, b in PAIRS])
SEVMISS = {nm: int((sev & ~REF[nm]).sum()) for nm in NAMES}
MV = {}
for a in NAMES:
    for b in NAMES:
        if a != b:
            e = refer(np.maximum(RK[a], RK[b])); MV[(a, b)] = int((sev & ~REF[a]).sum() - (sev & ~e).sum())
full = {}
for nm, fullname in ANCHOR_FULL.items():
    df = pd.read_csv(f'{PREDS}/eyepacs_predictions_fresh_{fullname}_v2.csv', dtype={'image_id': str})
    pc = next(c for c in ('predicted_probability','probability','prob') if c in df.columns)
    full[nm] = refer(df.drop_duplicates('image_id').set_index('image_id')[pc].astype(float).reindex(FR.image_id).to_numpy())
AP = list(itertools.combinations(ANCHOR_FULL, 2))
cheap = [OUT[((OUT.a == a) & (OUT.b == b)) | ((OUT.a == b) & (OUT.b == a))].comiss_sev.iloc[0] for a, b in AP]
fullv = [comiss(full[a], full[b], sev) for a, b in AP]
GATE = {'n_anchor_pairs': len(AP), 'spearman_cheap_vs_full_severe_comiss': round(float(spearmanr(cheap, fullv)[0]), 3)}
GATE['passed'] = GATE['spearman_cheap_vs_full_severe_comiss'] >= 0.6
json.dump(GATE, open(f'{ZOO_OUT}/calibration_gate_amendment1.json', 'w'), indent=2); print('CALIBRATION GATE:', GATE)
if not GATE['passed']: raise RuntimeError('CALIBRATION GATE FAILED: per the amended plan the pilot stops here. Send calibration_gate_amendment1.json.')
```

```python
# Cell 7: fingerprints (only models without a current fingerprint file, i.e. the retrained ones) (no labels). Trained: perturbation response vectors, occlusion maps, penultimate features. Pretrained: features and feature shifts.
def to_tensor(u8): return ((torch.from_numpy(u8).permute(0, 3, 1, 2).float() / 255.0) - MEAN) / STD
@torch.no_grad()
def run_batches(fn, arr, bs=100):
    out = []
    for s in range(0, len(arr), bs):
        with torch.amp.autocast('cuda'): out.append(fn(to_tensor(arr[s:s + bs]).to(dev)).float().cpu())
    return torch.cat(out).numpy().astype(np.float64)
OCC_N, GRID = 200, 8; P = 224 // GRID
@torch.no_grad()
def occlusion_maps(m):
    maps = np.zeros((OCC_N, GRID * GRID))
    for i in range(OCC_N):
        x = to_tensor(CLEAN[i:i + 1]).to(dev); base = score(m(x)).item()
        xs = x.repeat(GRID * GRID, 1, 1, 1)
        for k in range(GRID * GRID):
            r, c = divmod(k, GRID); xs[k, :, r * P:(r + 1) * P, c * P:(c + 1) * P] = 0
        with torch.amp.autocast('cuda'): maps[i] = base - score(m(xs)).float().cpu().numpy()
    return maps
def pretrained_feats(name, arr):
    m = build(name, pretrained=True, nc=0 if Z.loc[name, 'timm_id'] != 'retfound' else 5).to(dev).eval()
    if isinstance(m, RF): f = lambda x: m.base(pixel_values=x).last_hidden_state[:, 1:].mean(1)
    elif Z.loc[name, 'recipe'] == 'tfm' and hasattr(m, 'num_prefix_tokens') and not Z.loc[name, 'timm_id'].startswith(('swin', 'mixer')):
        npt = m.num_prefix_tokens; f = lambda x: m.forward_features(x)[:, npt:].mean(1)
    else: f = lambda x: m(x)
    out = {'clean': run_batches(f, arr)}
    for k in ('lowpass_r4', 'highpass_r4', 'drop_green', 'jpeg_q15', 'vignette'): out[k] = run_batches(f, PERT[k])
    del m; torch.cuda.empty_cache(); return out
for name in Z.index:
    fp = f'{ZOO_OUT}/fingerprints/{name}.npz'
    if os.path.exists(fp): continue
    t0 = time.time(); m = build(name, pretrained=False).to(dev); m.load_state_dict(torch.load(f'{ZOO_CK}/{name}.pt', map_location='cpu', weights_only=False)['state_dict']); m.eval()
    clean_s = run_batches(lambda x: score(m(x)), CLEAN)
    resp = {k: run_batches(lambda x: score(m(x)), PERT[k]) - clean_s for k in PERTURB}
    feats = run_batches(lambda x: penultimate(m, x), CLEAN); occ = occlusion_maps(m)
    del m; torch.cuda.empty_cache(); pt = pretrained_feats(name, CLEAN)
    pt_shift = {k: np.linalg.norm(pt[k] - pt['clean'], axis=1) / (np.linalg.norm(pt['clean'], axis=1) + 1e-8) for k in pt if k != 'clean'}
    np.savez_compressed(fp, clean_score=clean_s, feats=feats, occ=occ, pt_feats=pt['clean'],
                        **{f'resp_{k}': v for k, v in resp.items()}, **{f'ptshift_{k}': v for k, v in pt_shift.items()})
    print(f'{name:15s} fingerprints {time.time()-t0:.0f}s')
```

```python
# Cell 8: STAGE 5a, pair table of fingerprint similarities
FPX = {nm: dict(np.load(f'{ZOO_OUT}/fingerprints/{nm}.npz')) for nm in NAMES}
def cka(X, Y):
    X = X - X.mean(0); Y = Y - Y.mean(0); K, L = X @ X.T, Y @ Y.T
    return float((K * L).sum() / (np.linalg.norm(K) * np.linalg.norm(L)))
def rho(u, v): return float(spearmanr(u, v)[0])
GROUPS = {g: [k for k, (gg, _) in PERTURB.items() if gg == g] for g in ('frequency', 'channel', 'degradation')}
rows = []
for a, b in PAIRS:
    A, B = FPX[a], FPX[b]; r = {'a': a, 'b': b}
    for g, ks in GROUPS.items(): r[f'sim_{g}'] = np.mean([rho(A[f'resp_{k}'], B[f'resp_{k}']) for k in ks])
    r['sim_occlusion'] = np.nanmean([rho(A['occ'][i], B['occ'][i]) for i in range(OCC_N)])
    r['sim_clean_scores'] = rho(A['clean_score'], B['clean_score'])
    r['cka_trained'] = cka(A['feats'], B['feats']); r['cka_pretrained'] = cka(A['pt_feats'], B['pt_feats'])
    r['sim_pretrained_shift'] = np.mean([rho(A[k], B[k]) for k in A if k.startswith('ptshift_')])
    r['same_family'] = int(Z.loc[a, 'family'] == Z.loc[b, 'family']); r['same_data'] = int(Z.loc[a, 'pretrain_data'] == Z.loc[b, 'pretrain_data'])
    r['same_objective'] = int(Z.loc[a, 'objective'] == Z.loc[b, 'objective']); r['same_recipe'] = int(Z.loc[a, 'recipe'] == Z.loc[b, 'recipe'])
    r['mean_sevmiss'] = (SEVMISS[a] + SEVMISS[b]) / 2; r['gap_sevmiss'] = abs(SEVMISS[a] - SEVMISS[b])
    rows.append(r)
PT = pd.DataFrame(rows).merge(OUT, on=['a', 'b']); PT.to_csv(f'{ZOO_OUT}/pair_table.csv', index=False)
FPRED = ['sim_frequency','sim_channel','sim_degradation','sim_occlusion','cka_trained','cka_pretrained','sim_pretrained_shift','sim_clean_scores']
print(PT[FPRED + ['comiss_sev']].corr(method='spearman')['comiss_sev'].round(3).to_string())
```

```python
# Cell 9: strength-aware analysis
from sklearn.linear_model import RidgeCV
Y = 'comiss_sev'; COV = ['mean_sevmiss', 'gap_sevmiss']; SPEC = ['same_family', 'same_data', 'same_objective', 'same_recipe']
LABEL_FREE = [c for c in FPRED if c != 'sim_clean_scores']
D = PT.dropna(subset=[Y]).copy(); D['logY'] = np.log(D[Y]); rng = np.random.default_rng(3)
def partial_rho(df, p):
    Z_ = np.column_stack([np.ones(len(df)), rankdata(df.mean_sevmiss), rankdata(df.gap_sevmiss)])
    rx = rankdata(df[p]) - Z_ @ np.linalg.lstsq(Z_, rankdata(df[p]), rcond=None)[0]
    ry = rankdata(df[Y]) - Z_ @ np.linalg.lstsq(Z_, rankdata(df[Y]), rcond=None)[0]
    return float(np.corrcoef(rx, ry)[0, 1])
def boot(fn):
    bs = []
    for _ in range(500):
        keep = set(rng.choice(NAMES, len(NAMES), replace=True)); sub = D[D.a.isin(keep) & D.b.isin(keep)]
        if len(sub) > 15: bs.append(fn(sub))
    return [round(float(np.nanpercentile(bs, 2.5)), 3), round(float(np.nanpercentile(bs, 97.5)), 3)]
D['band'] = pd.qcut(D.mean_sevmiss, 3, labels=['strong_pairs', 'middle_pairs', 'weak_pairs'])
ANALYSIS = {}
for p in FPRED + SPEC:
    ANALYSIS[p] = {'raw_rho': round(rho(D[p], D[Y]), 3), 'raw_ci95': boot(lambda s: rho(s[p], s[Y])),
                   'partial_rho_given_strength': round(partial_rho(D, p), 3), 'partial_ci95': boot(lambda s: partial_rho(s, p)),
                   'within_band_rho': {b: round(rho(D[D.band == b][p], D[D.band == b][Y]), 3) for b in ('strong_pairs', 'middle_pairs', 'weak_pairs')}}
print(pd.DataFrame({p: {'raw': v['raw_rho'], 'partial': v['partial_rho_given_strength'], 'partial_ci': v['partial_ci95'], **v['within_band_rho']} for p, v in ANALYSIS.items()}).T.to_string())
def lofo(features):
    pred = pd.Series(np.nan, index=D.index)
    for fam in Z.family.unique():
        held = set(Z.index[Z.family == fam]); te = D.a.isin(held) | D.b.isin(held); tr = ~te
        if te.sum() == 0 or tr.sum() < 20: continue
        X = D[features].values; mu, sd = X[tr].mean(0), X[tr].std(0) + 1e-9
        mdl = RidgeCV(alphas=np.logspace(-3, 3, 13)).fit((X[tr] - mu) / sd, D.logY[tr]); pred[te] = mdl.predict((X[te] - mu) / sd)
    ok = pred.notna(); resid_obs = D.logY[ok] - pred[ok]
    return pred, round(rho(pred[ok], D.logY[ok]), 3)
MODELS_ = {'accuracy_only': COV, 'spec_sheet': COV + SPEC, 'fingerprints': COV + LABEL_FREE, 'fingerprints_plus_spec': COV + SPEC + LABEL_FREE, 'test_firing_only': COV + ['sim_clean_scores']}
LOFO = {k: lofo(v) for k, v in MODELS_.items()}
acc_pred = LOFO['accuracy_only'][0]; fp_pred = LOFO['fingerprints'][0]; ok = acc_pred.notna() & fp_pred.notna()
LOFO_INCREMENT = round(rho(fp_pred[ok] - acc_pred[ok], D.logY[ok] - acc_pred[ok]), 3)   # does the fingerprint model explain what accuracy leaves over?
print('LOFO rank correlation:', {k: v[1] for k, v in LOFO.items()}, '| increment over accuracy (residual rho):', LOFO_INCREMENT)
predF = LOFO['fingerprints'][0]
def pred_pair(a, b):
    r = D[((D.a == a) & (D.b == b)) | ((D.a == b) & (D.b == a))]; return float(predF[r.index[0]]) if len(r) and not np.isnan(predF[r.index[0]]) else np.inf
CLIN_OUT = {}
for tier_label, tol in (('primary_10pct', 1.10), ('secondary_25pct', 1.25)):
    rows = []
    for a in NAMES:
        cands = [b for b in NAMES if b != a]; best = min(SEVMISS[b] for b in cands); tier = [b for b in cands if SEVMISS[b] <= tol * best]
        strongest = min(cands, key=lambda b: SEVMISS[b]); pick = min(tier, key=lambda b: pred_pair(a, b))
        rows.append({'anchor': a, 'n_in_tier': len(tier), 'strongest': strongest, 'fingerprint_pick': pick, 'mv_strongest': MV[(a, strongest)], 'mv_fingerprint': MV[(a, pick)],
                     'oracle': max(MV[(a, b)] for b in cands)})
    CL = pd.DataFrame(rows); d = (CL.mv_fingerprint - CL.mv_strongest).to_numpy(); bs = [rng.choice(d, len(d), replace=True).mean() for _ in range(2000)]
    CLIN_OUT[tier_label] = {'mean_extra_severe_rescued_vs_strongest': round(float(d.mean()), 2), 'ci95': [round(float(np.percentile(bs, 2.5)), 2), round(float(np.percentile(bs, 97.5)), 2)],
                            'anchors_better': int((d > 0).sum()), 'anchors_worse': int((d < 0).sum()), 'anchors_same_pick': int((CL.fingerprint_pick == CL.strongest).sum()),
                            'mean_tier_size': round(float(CL.n_in_tier.mean()), 2), 'mean_oracle_minus_strongest': round(float((CL.oracle - CL.mv_strongest).mean()), 2),
                            'detail': CL.to_dict(orient='records')}
    print(tier_label, {k: v for k, v in CLIN_OUT[tier_label].items() if k != 'detail'})
```
